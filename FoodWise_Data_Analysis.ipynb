```python
# ==========================================
# FOODWISE - DATA CLEANING
# ==========================================

# Make a copy of the original dataset
df_clean = df.copy()

# 1. Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# 2. Handle missing numerical values
# Fill numerical missing values with the median
numeric_columns = df_clean.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df_clean[column] = df_clean[column].fillna(df_clean[column].median())

# 3. Handle missing categorical values
# Fill categorical missing values with the mode
categorical_columns = df_clean.select_dtypes(include="object").columns

for column in categorical_columns:
    if df_clean[column].isnull().sum() > 0:
        df_clean[column] = df_clean[column].fillna(df_clean[column].mode()[0])

# 4. Check whether missing values remain
print("===== MISSING VALUES AFTER CLEANING =====")
print(df_clean.isnull().sum())

# 5. Check duplicates again
print("\n===== DUPLICATES AFTER CLEANING =====")
print(df_clean.duplicated().sum())

# 6. Check final dataset size
print("\n===== FINAL DATASET SHAPE =====")
print(df_clean.shape)

print("\nData cleaning completed successfully!")
```

```python
# ==========================================
# FOODWISE - K-MEANS CLUSTERING
# ==========================================

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Use numerical features for clustering
cluster_data = df_clean.select_dtypes(include=np.number).copy()

# Standardize the data
scaler = StandardScaler()
cluster_scaled = scaler.fit_transform(cluster_data)

# Create K-Means model with 3 clusters
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Fit clustering model
clusters = kmeans.fit_predict(cluster_scaled)

# Add cluster labels to dataset
df_clean["Cluster"] = clusters

print("===== K-MEANS CLUSTERING RESULTS =====")
print(df_clean["Cluster"].value_counts().sort_index())

print("\nK-Means clustering completed successfully!")
```

```python
# ==========================================
# FOODWISE - PCA ANALYSIS
# ==========================================

from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Select numerical data
pca_data = df_clean.select_dtypes(include=np.number).copy()

# Remove Cluster column if it exists
if "Cluster" in pca_data.columns:
    pca_data = pca_data.drop(columns=["Cluster"])

# Standardize the data
pca_scaler = StandardScaler()
pca_scaled = pca_scaler.fit_transform(pca_data)

# Apply PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(pca_scaled)

# Create PCA DataFrame
pca_df = pd.DataFrame(
    pca_result,
    columns=["PC1", "PC2"]
)

# Add cluster information if available
if "Cluster" in df_clean.columns:
    pca_df["Cluster"] = df_clean["Cluster"].values

print("===== PCA RESULTS =====")
print("Variance explained by PC1:",
      round(pca.explained_variance_ratio_[0] * 100, 2), "%")

print("Variance explained by PC2:",
      round(pca.explained_variance_ratio_[1] * 100, 2), "%")

print("Total variance explained:",
      round(pca.explained_variance_ratio_.sum() * 100, 2), "%")

# PCA visualization
plt.figure(figsize=(8, 5))

if "Cluster" in pca_df.columns:
    sns.scatterplot(
        data=pca_df,
        x="PC1",
        y="PC2",
        hue="Cluster",
        palette="Set2",
        s=70
    )
else:
    sns.scatterplot(
        data=pca_df,
        x="PC1",
        y="PC2",
        s=70
    )

plt.title("PCA Visualization of Food Waste Data")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()

print("\nPCA analysis completed successfully!")
```
